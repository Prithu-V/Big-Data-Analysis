```sql
CREATE SCHEMA IF NOT EXISTS marketing_analytics.bronze;

CREATE OR REPLACE TABLE marketing_analytics.bronze.ab_test_raw
TBLPROPERTIES (
  'delta.columnMapping.mode' = 'name',
  'delta.minReaderVersion' = '2',
  'delta.minWriterVersion' = '5'
)
AS
SELECT *
FROM read_files(
  '/Volumes/marketing_analytics/raw/landing/marketing_AB.csv',
  format => 'csv',
  header => true,
  inferColumnTypes => false
);
```

```sql
SELECT COUNT(*) AS row_count FROM marketing_analytics.bronze.ab_test_raw;

SELECT * FROM marketing_analytics.bronze.ab_test_raw LIMIT 10;
```

```sql
SELECT
  COUNT(*)                   AS total_rows,
  COUNT(DISTINCT `user id`)  AS distinct_users
FROM marketing_analytics.bronze.ab_test_raw;
```

```sql
SELECT
  COUNT_IF(`user id`       IS NULL OR `user id`       = '') AS null_user_id,
  COUNT_IF(`test group`    IS NULL OR `test group`    = '') AS null_test_group,
  COUNT_IF(`converted`     IS NULL OR `converted`     = '') AS null_converted,
  COUNT_IF(`total ads`     IS NULL OR `total ads`     = '') AS null_total_ads,
  COUNT_IF(`most ads day`  IS NULL OR `most ads day`  = '') AS null_day,
  COUNT_IF(`most ads hour` IS NULL OR `most ads hour` = '') AS null_hour
FROM marketing_analytics.bronze.ab_test_raw;
```

```sql
SELECT
  `test group`,
  COUNT(*) AS users,
  ROUND(100.0 * COUNT(*) / SUM(COUNT(*)) OVER (), 2) AS pct_of_total
FROM marketing_analytics.bronze.ab_test_raw
GROUP BY `test group`;
```

```sql
SELECT DISTINCT `converted`     FROM marketing_analytics.bronze.ab_test_raw;
SELECT DISTINCT `most ads day`  FROM marketing_analytics.bronze.ab_test_raw;
SELECT MIN(CAST(`most ads hour` AS INT)) AS min_hour,
       MAX(CAST(`most ads hour` AS INT)) AS max_hour
FROM marketing_analytics.bronze.ab_test_raw;
```
