```sql
CREATE SCHEMA IF NOT EXISTS marketing_analytics.silver;

CREATE OR REPLACE TABLE marketing_analytics.silver.ab_test_clean AS
SELECT
  CAST(`user id` AS BIGINT)        AS user_id,
  `test group`                     AS test_group,
  CASE
    WHEN `converted` = 'True'  THEN TRUE
    WHEN `converted` = 'False' THEN FALSE
  END                              AS converted,
  CAST(`total ads` AS INT)         AS total_ads,
  `most ads day`                   AS most_ads_day,
  CAST(`most ads hour` AS INT)     AS most_ads_hour
FROM marketing_analytics.bronze.ab_test_raw;
```

```sql
SELECT
  COUNT(*)                                AS total_rows,
  COUNT_IF(converted IS NULL)             AS null_converted,
  MIN(total_ads)                          AS min_ads,
  MAX(total_ads)                          AS max_ads
FROM marketing_analytics.silver.ab_test_clean;
```

```sql
DESCRIBE marketing_analytics.silver.ab_test_clean;
```

```sql
SELECT
  PERCENTILE(total_ads, 0.50) AS median_ads,
  PERCENTILE(total_ads, 0.90) AS p90_ads,
  PERCENTILE(total_ads, 0.99) AS p99_ads,
  PERCENTILE(total_ads, 0.999) AS p999_ads,
  MAX(total_ads)              AS max_ads
FROM marketing_analytics.silver.ab_test_clean;
```
